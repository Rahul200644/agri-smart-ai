```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectFromModel
from sklearn.linear_model import Lasso
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report

import joblib
```

```python
df = pd.read_csv("/content/startup data (2).csv")

print(df.head())
print(df.shape)
```

```python
print(df.info())

print(df.describe())

print(df.isnull().sum())

print(df["status"].value_counts())
```

```python
for col in df.columns:

    if df[col].dtype=="object":

        df[col]=df[col].fillna("Unknown")

    else:

        df[col]=df[col].fillna(df[col].median())
```

```python
print(df.isnull().sum())
```

```python
le=LabelEncoder()

for col in df.columns:

    if df[col].dtype=="object":

        df[col]=le.fit_transform(df[col].astype(str))
```

```python
df["status"]=df["status"].map({
    le.transform(["acquired"])[0]:1,
    le.transform(["closed"])[0]:0
})

print(df["status"].value_counts())
```

```python
X=df.drop("status",axis=1)

y=df["status"]

print(X.shape)

print(y.shape)
```

```python
scaler=StandardScaler()

X_scaled=scaler.fit_transform(X)

lasso=Lasso(alpha=0.01)

lasso.fit(X_scaled,y)

selector=SelectFromModel(lasso,prefit=True)

X_selected=selector.transform(X_scaled)

print("Original Features :",X.shape[1])

print("Selected Features :",X_selected.shape[1])
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_selected,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("Training Data Shape:", X_train.shape)
print("Testing Data Shape:", X_test.shape)
```

```python
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)

print("Model Trained Successfully!")
```

```python
y_pred = model.predict(X_test)

print("First 10 Predictions:")
print(y_pred[:10])
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy :", accuracy)
print("Accuracy Percentage :", accuracy*100)
```

```python
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix")
print(cm)
```

```python
selected_features = X.columns[selector.get_support()]

importance = pd.Series(
    model.feature_importances_,
    index=selected_features
)

importance = importance.sort_values(ascending=False)

print(importance)
```

```python
plt.figure(figsize=(10,6))

importance.head(10).plot(kind="bar")

plt.title("Top 10 Important Features")

plt.xlabel("Features")

plt.ylabel("Importance")

plt.show()
```

```python
sample = X_selected[0].reshape(1, -1)

prediction = model.predict(sample)

if prediction[0] == 1:
    print("Prediction : Startup will be Successful (Acquired)")
else:
    print("Prediction : Startup may Fail (Closed)")
```

```python
joblib.dump(model, "startup_success_model.pkl")

print("Model Saved Successfully!")
```
